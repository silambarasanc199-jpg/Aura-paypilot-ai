```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import requests
import zipfile
import io

sns.set_theme(style="whitegrid")

print("VINIX Task 2 - Student Performance Analysis")
print("All libraries imported successfully!")
```

```python
url = "https://archive.ics.uci.edu/static/public/320/student+performance.zip"

response = requests.get(url)
response.raise_for_status()

# Open the outer zip file
outer_zip = zipfile.ZipFile(io.BytesIO(response.content))

# Open the inner zip file 'student.zip' which contains the CSVs
with outer_zip.open("student.zip") as inner_zip_file:
    inner_zip = zipfile.ZipFile(io.BytesIO(inner_zip_file.read()))
    with inner_zip.open("student-mat.csv") as file:
        df = pd.read_csv(file, sep=";")

print("Dataset loaded successfully!")
print("Number of records:", len(df))
print("Number of columns:", len(df.columns))

display(df.head())
```

```python

print("DATASET INFORMATION")
print("--------------------")

df.info()

print("\nSTATISTICAL SUMMARY")
display(df.describe())

print("\nMISSING VALUES")
display(df.isnull().sum())

print("\nDUPLICATE RECORDS")
print(df.duplicated().sum())

print("\nCOLUMN NAMES")
print(df.columns.tolist())
```

```python

df_clean = df.copy()

# Remove duplicate records
df_clean = df_clean.drop_duplicates()

# Handle missing numerical values
for col in df_clean.select_dtypes(include="number").columns:
    df_clean[col] = df_clean[col].fillna(
        df_clean[col].median()
    )

# Handle missing categorical values
for col in df_clean.select_dtypes(include="object").columns:
    if df_clean[col].isnull().any():
        df_clean[col] = df_clean[col].fillna(
            df_clean[col].mode()[0]
        )

print("Data cleaning completed!")
print("Remaining missing values:",
      df_clean.isnull().sum().sum())
print("Final dataset shape:", df_clean.shape)
```

```python

print("STUDENT PERFORMANCE ANALYSIS")
print("=============================")

print("Total students:", len(df_clean))
print("Average final grade:", round(df_clean["G3"].mean(), 2))
print("Highest grade:", df_clean["G3"].max())
print("Lowest grade:", df_clean["G3"].min())

print("\nAverage grade by study time:")
display(
    df_clean.groupby("studytime")["G3"].mean().round(2)
)

print("\nAverage grade by previous failures:")
display(
    df_clean.groupby("failures")["G3"].mean().round(2)
)

print("\nAverage grade by gender:")
display(
    df_clean.groupby("sex")["G3"].mean().round(2)
)
```

```python

plt.figure(figsize=(8, 5))
sns.histplot(df_clean["G3"], bins=15, kde=True)
plt.title("Distribution of Student Final Grades")
plt.xlabel("Final Grade")
plt.ylabel("Number of Students")
plt.show()
```

```python

plt.figure(figsize=(8, 5))
sns.boxplot(data=df_clean, x="studytime", y="G3")
plt.title("Study Time vs Final Grade")
plt.xlabel("Study Time Category")
plt.ylabel("Final Grade")
plt.show()
```

```python

plt.figure(figsize=(8, 5))
sns.scatterplot(data=df_clean, x="absences", y="G3")
plt.title("Student Absences vs Final Grade")
plt.xlabel("Absences")
plt.ylabel("Final Grade")
plt.show()
```

```python

print("KEY FINDINGS")
print("============")

print("Total students:", len(df_clean))
print("Average final grade:", round(df_clean["G3"].mean(), 2))

print("Students scoring 10 or above:",
      int((df_clean["G3"] >= 10).sum()))

print("Students scoring below 10:",
      int((df_clean["G3"] < 10).sum()))

print("Study time and final grade correlation:",
      round(df_clean["studytime"].corr(df_clean["G3"]), 3))

print("Absences and final grade correlation:",
      round(df_clean["absences"].corr(df_clean["G3"]), 3))

print("\nAnalysis completed successfully!")
```

```python

df_clean.to_csv(
    "student_performance_cleaned.csv",
    index=False
)

print("Cleaned dataset saved successfully!")
```
